## Data Cleaning

### Remove indexing in Movie_Title

```python
# remove numbering from the Movie_Title column
movies_df['Movie_Title'] = movies_df['Movie_Title'].str.replace(r'^\d+\.\s*', '', regex=True)
movies_df.head()
```

### Clean Storyline column

```python
import re

def clean_text(text):

    text = str(text)

    # Convert to lowercase
    text = text.lower()

    # Remove special characters
    text = re.sub(r"[^a-zA-Z\s]", " ", text)

    # Remove extra spaces
    text = re.sub(r"\s+", " ", text).strip()

    return text


movies_df["clean_storyline"] = movies_df["Storyline"].apply(clean_text)
movies_df.tail()
```

```python
# save the cleaned dataframe to a new CSV file
movies_df.to_csv('../data/imdb_data_cleaned.csv', index=False)
```
